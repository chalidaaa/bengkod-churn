## 3. Aplikasi Streamlit

File `deployment/app.py` membangun antarmuka yang berisi:
- **Pemuatan model** — `joblib.load` dengan cache.

- **Tampilan hasil** — label churn/bertahan beserta probabilitasnya.

Menjalankan secara lokal:
```bash

```

## 4. Deployment ke Streamlit Cloud

1. Siapkan `requirements.txt` (versi paket disamakan dengan versi saat model dibuat).
2. Unggah proyek ke GitHub (termasuk `deployment/churn_model.joblib`).
3. Buka [share.streamlit.io](https://share.streamlit.io), hubungkan repo.
4. Set **Main file path** = `deployment/app.py`, lalu **Deploy**.
5. Aplikasi dapat diakses publik via URL Streamlit Cloud.

Detail langkah ada di `deployment/README.md`.
